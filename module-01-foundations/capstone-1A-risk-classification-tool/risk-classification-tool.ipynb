**EU AI Act Risk Classification Tool**


```python
classification_rules = {
    "A": {"risk_tier":"High-Risk  + Regulatory Gap",
        "eu_ai_act_basis": "Article 52 + Article 5(1)(e)"
    },
    "B":{"risk_tier": "Unacceptable Risk",
        "eu_ai_act_basis": "Article 5"
        },
    "C":{"risk_tier": "High-Risk",
        "eu_ai_act_basis": "Annex III"
        },
     "D":{"risk_tier": "High-Risk",
        "eu_ai_act_basis": "Annex III"
        }
}
```

```python
def get_user_input():
    purpose = " "
    while purpose not in ["A", "B", "C", "D"]:
        print("What is the primary purpose of the system?")
        print("A) Generate synthetic media/images")
        print("B) Identify or track people in real-time")
        print("C) Screen or rank candidates for employment")
        print("D) Moderate user content")
        purpose = input("Enter your choice (A/B/C/D): ").upper()
    if purpose not in ["A", "B", "C", "D"]:
        print("Invalid answer, please try again")
 

    print("Does it process biometric data?")
    Biometric_data=input("Yes or no ")
    print("Does it operate in public spaces in real-time?")
    operate_inpublic=input("Yes or no ")

    print("Who is the affected population?")
    print("A) General public")
    print("B) Job applicants ")
    print("C) Specific individuals whose likeness is used")
    affected_pop=input("Enter your choice (A/B/C):")
    
    print("Does it make consequential decisions affecting fundamental rights?")
    descion_forrights=input("Yes or no ")

    
    return purpose,Biometric_data,operate_inpublic, affected_pop,descion_forrights
```

```python
def classify_system():
    val = get_user_input()[0].upper()
    ans = classification_rules[val]
    one= (ans.get("risk_tier")) 
    two= (ans.get("eu_ai_act_basis"))  
    return print(f"""=== EU AI Act Risk Classification ===
Risk Tier: {one}
EU AI Act Basis: {two}
=====================================""")

    
```

```python
classify_system()
```
