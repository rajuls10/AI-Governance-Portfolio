# Cluster 3 — Governance Concepts Summary

## Concept Rankings by Governance Impact

| Concept | Governance Impact | When to Use | Real-Life Example | Frameworks |
|---------|------------------|-------------|-------------------|------------|
| **Mean & Standard Deviation** | Medium — describes data but doesn't prove bias | Initial data audit, baseline reporting, comparing group averages | A bank shows men and women have identical average credit scores. Looks fair. But tells you nothing about rejection rates or outcome fairness. | NIST RMF Map (understand your system) |
| **Hypothesis Testing (p-value)** | High — proves whether bias is statistically real or random | When you need to prove bias exists to a regulator or court | A hiring AI shows women score 5 points lower. Is it real? p-value < 0.05 confirms it's statistically significant — the company must explain it. | NIST RMF Measure, EU AI Act Annex III audit requirements |
| **Disparate Impact (80% Rule)** | High — catches hidden bias in outcomes even when scoring is fair | After hiring/lending decisions — check if outcomes are equal across groups | A hiring system scores fairly but only 40% of women get hired vs 52% of men. 76.9% ratio triggers the 80% rule — investigation required. | EU AI Act Annex III, NIST RMF Manage, CEDAW Article 11 (employment discrimination) |
| **Probability Analysis** | Very High — translates failure rates into real human harm at scale | Risk assessments, policy advocacy, regulatory standard-setting | ContentGuard fails 30% of the time. Sounds small. But 27 women per day have all NCII reports ignored. Required failure rate: <17.1%. | EU AI Act Article 52, NIST RMF Measure + Manage, DPDP Act fairness principle |
