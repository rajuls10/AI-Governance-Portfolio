# Exercise 02: Automated governance profile generator

```python
def print_governance_profile(system_name):
    for system in ai_systems:
        if system['name'] == system_name:
            print(f"GOVERNANCE PROFILE: {system['name']}")
            print(f"Purpose: {system['purpose']}")
            print(f"Affected Population: {system['affected_population']}")
            print(f"Risk Tier: {system['risk_tier']}")
            print(f"EU AI Act Basis: {system['eu_ai_act_basis']}")
            print(f"Case Study: {system['case_study']}")
            return
    print(f"System '{system_name}' not found in inventory.")
```

# Exercise 03: Explaining EU AI Act Basis

```python
list_euai_basis = []
for basis in ai_systems:
    list_euai_basis.append(basis["eu_ai_act_basis"])
print(list_euai_basis)
```

```python
def explain_eu_ai_act_basis(basis_name):
    explanations = {
        "Article 52 (transparency for AI-generated content) + Article 5(1)(e) (prohibited manipulation). NOTE: NCII generation is a regulatory gap in the EU AI Act — the framework does not explicitly classify generative image platforms.": "Article 52 requires AI-generated content including deepfakes to be labelled as artificial. Article 5(1)(e) prohibits AI that manipulates people in ways that harm their dignity or safety. SynthVision sits at the intersection of both — but the EU AI Act has a critical gap: it was designed to classify systems that identify or decide, not systems that generate. NCII generation requires governance beyond what current risk tiers cover.",
        "Article 5 - Prohibited real-time biometric surveillance in public spaces": "Using AI to scan and identify people live in public spaces is largely prohibited. Mass surveillance without consent or judicial oversight is incompatible with fundamental rights. Narrow law enforcement exceptions exist but require prior authorisation and strict safeguards.",
        "Annex III - Employment and workers management systems": "AI used in hiring, performance evaluation, promotion, or dismissal is high-risk. These systems directly affect people's livelihoods and are known to encode historical discrimination — particularly against women, minorities, and those with non-linear career paths.",
        "Annex III - Consequential content moderation decisions affecting users' fundamental rights. Dismissal of NCII reports constitutes a high-risk governance failure.": "Content moderation systems that make binding decisions — removing posts, dismissing abuse reports, restricting accounts — affect users' fundamental rights. When ContentGuard fails to act on an NCII report, that is not a minor system error. It is a governance failure with direct, measurable harm to a real person.",
        "Annex III - Access to essential private services and public benefits": "AI systems that determine access to credit, insurance, housing, or welfare are high-risk because a wrong decision can exclude people from essential economic participation. These systems often use proxy variables that correlate with race, gender, or class — encoding discrimination without naming it."
    }
    
    if basis_name in explanations:
        print(f"{basis_name}:\n{explanations[basis_name]}\n")
    else:
        print(f"'{basis_name}' not found in explainer.")
```

```python
explain_eu_ai_act_basis("Article 52 (transparency for AI-generated content) + Article 5(1)(e) (prohibited manipulation). NOTE: NCII generation is a regulatory gap in the EU AI Act — the framework does not explicitly classify generative image platforms.")
```

```python
for name in list_euai_basis:
    explain_eu_ai_act_basis(name)
```

# Exercise 04: AI Governance Report

```python
def generate_governance_report():
    high_risk=[]
    unacceptable_risk=[]
    limited_risk=[]
    for system in ai_systems:
        if system['risk_tier']=='High-Risk':
            high_risk.append(system['name'])
        elif system['risk_tier']=='Unacceptable Risk':
            unacceptable_risk.append(system['name'])
        elif system['risk_tier']=='Limited Risk':
            limited_risk.append(system['name'])
    print("========================================")
    print("AI GOVERNANCE REPORT")
    print("========================================")
    print(f"Total systems: {len(ai_systems)}")
    print(f"High-Risk Systems: {len(high_risk)}")
    print(f"Unacceptable Risk Systems: {len(unacceptable_risk)}")
    print(f"Limited Risk Systems: {len(limited_risk)}")
    print("")
    print("HIGH RISK SYSTEMS:")
    for risk in high_risk:
        print(f"  - {risk}")
    print("")
    print("UNACCEPTABLE RISK SYSTEMS:")
    for risk in unacceptable_risk:
        print(f"  - {risk}")
    print("")
    print("LIMITED RISK SYSTEMS:")
    for risk in limited_risk:
        print(f"  - {risk}")
    print("========================================")
```

```python
generate_governance_report()
```
