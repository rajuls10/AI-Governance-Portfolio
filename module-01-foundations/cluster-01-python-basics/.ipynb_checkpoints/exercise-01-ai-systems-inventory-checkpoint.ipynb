```python
# Exercise 01: AI Systems Inventory
# Cluster 1 - Python Basics
# Framework: EU AI Act Risk Classification
# Case Studies: NCII, Biometric Surveillance, Recruitment AI

# Each AI system is stored as a dictionary with key governance attributes

ai_systems = [
    {
        "name": "SynthVision AI",
        "purpose": "AI image generation platform capable of face-swapping and synthetic portrait generation",
        "data_inputs": "User-uploaded photos, text prompts",
        "affected_population": "Anyone whose likeness can be used without consent, primarily women",
        "risk_tier": "High-Risk",
        "eu_ai_act_basis": "Article 52 (transparency for AI-generated content) + Article 5(1)(e) (prohibited manipulation). NOTE: NCII generation is a regulatory gap in the EU AI Act.",
        "case_study": "NCII"
    },
    {
        "name": "FaceWatch Pro",
        "purpose": "Real-time facial recognition for law enforcement surveillance",
        "data_inputs": "Live CCTV footage, biometric database",
        "affected_population": "General public, disproportionately affects minorities",
        "risk_tier": "Unacceptable Risk",
        "eu_ai_act_basis": "Article 5 - Prohibited real-time biometric surveillance in public spaces",
        "case_study": "Biometric Surveillance"
    },
    {
        "name": "HireScore AI",
        "purpose": "Automated CV screening and candidate ranking for recruitment",
        "data_inputs": "CVs, LinkedIn profiles, assessment scores",
        "affected_population": "Job applicants, disproportionately affects women and minorities",
        "risk_tier": "High-Risk",
        "eu_ai_act_basis": "Annex III - Employment and workers management systems",
        "case_study": "Recruitment AI"
    },
    {
        "name": "ContentGuard",
        "purpose": "Automated content moderation for social media platforms",
        "data_inputs": "User posts, images, videos",
        "affected_population": "All platform users, particularly targets of gendered harassment",
        "risk_tier": "High-Risk",
       "eu_ai_act_basis": "Annex III - Consequential content moderation decisions affecting users fundamental rights.",
        "case_study": "NCII"
    },
    {
        "name": "CreditSense",
        "purpose": "AI-based credit scoring for loan applications",
        "data_inputs": "Financial history, spending patterns, demographic proxies",
        "affected_population": "Loan applicants, disproportionately affects low-income groups",
        "risk_tier": "High-Risk",
        "eu_ai_act_basis": "Annex III - Access to essential private services and public benefits",
        "case_study": "Recruitment AI"
    }
]

print(f"Total AI systems in inventory: {len(ai_systems)}")
print("Inventory loaded successfully.")
```

```python
# Print name and risk tier of each system
for system in ai_systems:
    print(f"{system['name']} — {system['risk_tier']}")
```

# Exercise 02: Automated governance profile generator

```python
def print_governance_profile(system_name):
    for system in ai_systems:
        if system['name'] == system_name:
            print(f"GOVERNANCE PROFILE: {system['name']}")
            print(f"Purpose: {system['purpose']}")
            print(f"Affected Population: {system['affected_population']}")
            print(f"Risk Tier: {system['risk_tier']}")
            print(f"EU AI Act Basis: {system['eu_ai_act_basis']}")
            print(f"Case Study: {system['case_study']}")
            return
    print(f"System '{system_name}' not found in inventory.")
```

```python
print_governance_profile("SynthVision AI")
```

```python
system_list=["FaceWatch Pro","HireScore AI"]
print_governance_profile(system_list[0])
```

```python
print_governance_profile("RandomAI")
```

```python
for system in ai_systems:
    print_governance_profile(system['name'])
```

# Exercise 03: Explaining Eu Ai Act basis 

```python
list_euai_basis = []
for basis in ai_systems:
    list_euai_basis.append(basis["eu_ai_act_basis"])
print(list_euai_basis)
```

```python
def explain_eu_ai_act_basis(basis_name):
    explanations = {
        "Article 52 (transparency for AI-generated content) + Article 5(1)(e) (prohibited manipulation). NOTE: NCII generation is a regulatory gap in the EU AI Act."
        "Article 5 - Prohibited real-time biometric surveillance in public spaces": "Using AI to scan and track people live in public places (like streets or stations) is mostly banned to protect freedom and privacy.",
        "Annex III - Employment and workers management systems": "AI used for hiring, promotions, firing, or monitoring workers is high-risk because it can strongly impact people's jobs and careers.",
        "Annex III - Consequential content moderation decisions affecting users fundamental rights."
        "Annex III - Access to essential private services and public benefits": "AI used in important services like banking, insurance, education, or welfare is high-risk because unfair decisions can seriously affect people's lives."
    }
    
    if basis_name in explanations:
        print(f"{basis_name}:\n{explanations[basis_name]}")
    else:
        print(f"'{basis_name}' not found.")
    
```

```python
for name in list_euai_basis:
    explain_eu_ai_act_basis(name)
```

```python
    print(f"total systems: {len(ai_systems)}")
```

```python
def generate_governance_report():
    high_risk=[]
    unacceptable_risk=[]
    limited_risk=[]
    for system in ai_systems:
        if system['risk_tier']=='High-Risk':
            high_risk.append(system['name'])
        elif system['risk_tier']=='Unacceptable Risk':
             unacceptable_risk.append(system['name'])
        elif system['risk_tier']=='Limited Risk':
            limited_risk.append(system['name'])
    print(f"Total systems: {len(ai_systems)}")
    print(f"High-Risk: {len(high_risk)}")
    print(f"Unacceptable Risk Systems {len(unacceptable_risk)}")
    print(f"Limited Risk Systems: {len(limited_risk)}")
    print("HIGH RISK SYSTEMS:")
    for risk in high_risk:
        print(f"- {risk}")

    print("UNACCEPTABLE RISK SYSTEMS:")   
    for risk in unacceptable_risk:
       print(f"- {risk}")

    print ("LIMITED RISK SYSTEMS:") 
    for risk in limited_risk:
       print(f"- {risk}")


    
```

```python
   generate_governance_report()
```
